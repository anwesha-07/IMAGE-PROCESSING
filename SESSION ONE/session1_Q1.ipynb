```python
import cv2
import matplotlib.pyplot as plt

image = cv2.imread(r"D:\image_processing\lena.jpg", cv2.IMREAD_GRAYSCALE)

if image is None:
    print("Error: Image not found.")
else:
    height, width = image.shape

    print("Image Height:", height)
    print("Image Width:", width)
    print("Number of Pixels:", image.size)
    print("Datatype:", image.dtype)
    print("Minimum Pixel Intensity:", image.min())
    print("Maximum Pixel Intensity:", image.max())
    print("Mean Intensity:", image.mean())

    x = 100
    y = 100

    print(f"Pixel value at ({x}, {y}):", image[y, x])

    # Display image in Jupyter
    
    plt.imshow(image, cmap="gray")
    plt.title("Grayscale Image")
    plt.axis("off")
    plt.show()
```
