```python
import cv2
import numpy as np
import matplotlib as plt

# Read grayscale image
image = cv2.imread(r"D:\image_processing\lena.jpg", cv2.IMREAD_GRAYSCALE)

# Check whether image is loaded
if image is None:
    print("Error: Image not found.")
else:
    print("Image loaded successfully.")
    print("Image shape:", image.shape)
```

```python
plt.imshow(image, cmap="gray")
plt.title("Grayscale Image")
plt.axis("off")
plt.show()
```

```python
brightness_increased = image + 50
plt.imshow(brightness_increased, cmap="gray")
plt.title("Increased Brightness")
plt.axis("off")
plt.show()
```

```python
brightness_decreased = image-50
plt.imshow(brightness_increased, cmap="gray")
plt.title("Decreased Brightness")
plt.axis("off")
plt.show()
```

```python
contrast_increased = 128 + 2 * (image - 128)

plt.imshow(contrast_increased, cmap="gray")
plt.title("Increased Contrast")
plt.axis("off")
plt.show()
```

```python
contrast_decreased = 128 + 0.5 * (image - 128)

plt.imshow(contrast_decreased, cmap="gray")
plt.title("Decreased Contrast")
plt.axis("off")
plt.show()
```

```python
negative = 255 - image

plt.imshow(negative, cmap="gray")
plt.title("Negative Image")
plt.axis("off")
plt.show()
```

```python
threshold = 128

binary = np.where(image >= threshold, 255, 0)

plt.imshow(binary, cmap="gray")
plt.title("Binary Threshold")
plt.axis("off")
plt.show()
```
