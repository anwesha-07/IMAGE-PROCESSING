```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
image = cv2.imread("lena.jpg", 0)

plt.imshow(image, cmap="gray")
plt.title("Lena Image")
plt.axis("off")
plt.show()
```

```python
dark = image * 0.4
dark = dark.astype(np.uint8)
```

```python
hist = cv2.calcHist([dark], [0], None, [256], [0, 256])
```

```python
plt.plot(hist)
plt.title("Histogram of Dark Image")
plt.xlabel("Intensity")
plt.ylabel("Number of Pixels")
plt.show()
```
