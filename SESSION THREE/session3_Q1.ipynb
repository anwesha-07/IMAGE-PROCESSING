```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

image = cv2.imread("lena.jpg", 0)

plt.imshow(image, cmap="gray")
plt.title("Original Lena Image")
plt.show()
```

# 4-neighbour connectivity

```python
pixel = (2, 2)

r, c = pixel

neighbors_4 = [
    (r-1, c),
    (r+1, c),
    (r, c-1),
    (r, c+1)
]

print("4-neighbours:", neighbors_4)
```

# 8-neighbours connectivity

```python
neighbors_8 = [
    (r-1, c-1), (r-1, c), (r-1, c+1),
    (r, c-1),             (r, c+1),
    (r+1, c-1), (r+1, c), (r+1, c+1)
]

print("8-neighbours:", neighbors_8)
```

# m-negighbours connectivity

```python
neighbors_m = []

# 4-neighbours are always considered
neighbors_m.extend([
    (r-1, c),
    (r+1, c),
    (r, c-1),
    (r, c+1)
])

# Check diagonal neighbours
diagonal = [
    (r-1, c-1),
    (r-1, c+1),
    (r+1, c-1),
    (r+1, c+1)
]

for p in diagonal:
    if image[p] == 1:
        neighbors_m.append(p)

print("m-neighbours:", neighbors_m)
```

# Euclidean Distance

```python
p1 = (1, 1)
p2 = (3, 3)

euclidean = np.sqrt((p1[0]-p2[0])**2 + (p1[1]-p2[1])**2)

print("Euclidean Distance:", euclidean)
```

# city bock distance

```python
city_block = abs(p1[0]-p2[0]) + abs(p1[1]-p2[1])

print("City-Block Distance:", city_block)
```

# Chessboard Distance

```python
chessboard = max(abs(p1[0]-p2[0]), abs(p1[1]-p2[1]))

print("Chessboard Distance:", chessboard)
```
