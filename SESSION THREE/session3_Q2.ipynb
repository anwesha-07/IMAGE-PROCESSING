```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

image = cv2.imread("lena.jpg", 0)

plt.imshow(image, cmap="gray")
plt.title("Original Lena Image")
plt.show()
```

# Calculate 2-D DFT

```python
dft = np.fft.fft2(image)

print("DFT calculated successfully.")
```

# Shift the Zero-Frequency Component to the Center

```python
dft_shift = np.fft.fftshift(dft)
```

# Calculate Magnitude Spectrum

```python
magnitude = np.abs(dft_shift)

plt.imshow(magnitude, cmap='gray')
plt.title("Magnitude Spectrum")
plt.axis("off")
plt.show()
```

# Logarithmic Scaling

```python
log_magnitude = np.log(1 + magnitude)

plt.imshow(log_magnitude, cmap='gray')
plt.title("Logarithmic Magnitude Spectrum")
plt.axis("off")
plt.show()
```

# Calculate 2-D DCT

```python
dct = cv2.dct(np.float32(image))
```

```python
print("DCT Coefficient Matrix:")
print(dct)
```

```python
plt.imshow(np.log(1 + np.abs(dct)), cmap='gray')
plt.title("DCT Coefficients")
plt.axis("off")
plt.show()
```
