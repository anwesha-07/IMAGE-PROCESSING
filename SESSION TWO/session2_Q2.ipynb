```python
import numpy as np
import cv2
import matplotlib.pyplot as plt
image = cv2.imread(r"D:\image_processing\lena.jpg", cv2.IMREAD_GRAYSCALE)

if image is None:
    print("Error: Image not found.")

plt.imshow(image, cmap="gray")
plt.title("Grayscale Image")
plt.axis("off")
plt.show()
```

```python
zoom2_nearest=cv2.resize(image,None,fx=2.0,fy=2.0,interpolation=cv2.INTER_NEAREST)
plt.imshow(zoom2_nearest, cmap="gray")
plt.show()
```

```python
zoom2_bilinear=cv2.resize(image,None,fx=2.0,fy=2.0,interpolation=cv2.INTER_LINEAR)
plt.imshow(zoom2_bilinear, cmap="gray")
plt.show()
```

```python
zoom2_bicubic=cv2.resize(image,None,fx=2.0,fy=2.0,interpolation=cv2.INTER_CUBIC)
plt.imshow(zoom2_bicubic, cmap="gray")
plt.show()
```
