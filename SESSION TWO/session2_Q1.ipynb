```python
import cv2
import matplotlib.pyplot as plt
import numpy as np

image = cv2.imread(r"D:\image_processing\lena.jpg", cv2.IMREAD_GRAYSCALE)

if image is None:
    print("Error: Ima not found.")

plt.imshow(image, cmap="gray")
plt.title("Grayscale Image")
plt.axis("off")
plt.show()
```

```python
height,width=image.shape[:2]
t=np.float32([[1, 0, 100],[0, 1, 50]])
translated=cv2.warpAffine(image,t,(width,height))
plt.imshow(translated,cmap="gray")
plt.show()
```

```python
resized_image=cv2.resize(image,None,fx=1.5,fy=1.5,interpolation=cv2.INTER_LINEAR)
plt.imshow(resized_image,cmap="gray")
plt.show()
```

```python
resized_image=cv2.resize(image,None,fx=1.5,fy=1.5,interpolation=cv2.INTER_AREA)
plt.imshow(resized_image,cmap="gray")
plt.show()
```

```python
resized_image=cv2.resize(image,None,fx=0.5,fy=0.5,interpolation=cv2.INTER_CUBIC)
plt.imshow(resized_image,cmap="gray")
plt.show()
```

```python
center=(width//2,height//2)
rotation_matrix=cv2.getRotationMatrix2D(center,45,1.0)
rotated_image=cv2.warpAffine(image,rotation_matrix,(width,height))
plt.imshow(rotated_image,cmap="gray")
plt.show()
```

```python
horizontal_image=cv2.flip(image,1)
vertical_image=cv2.flip(image,0)
plt.imshow(horizontal_image,cmap="gray")
plt.show()
```
